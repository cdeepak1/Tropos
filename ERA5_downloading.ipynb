```python
import cdsapi

c = cdsapi.Client()

years = range(2016, 2026)   # 2016–2025

for y in years:
    year = str(y)
    print(f"Requesting ERA5 total_precipitation for {year} ...")

    c.retrieve(
        "reanalysis-era5-single-levels",
        {
            "product_type": "reanalysis",
            "variable": ["total_precipitation"],
            "year": year,
            "month": [f"{m:02d}" for m in range(1, 13)],
            "day": [f"{d:02d}" for d in range(1, 32)],
            "time": [f"{h:02d}:00" for h in range(24)],
            # [North, West, South, East] – your region
            "area": [5.0, -75.0, -20.0, -35.0],
            "format": "netcdf",
        },
        f"era5_tp_region_{year}.nc",
    )

print("Done: downloaded ERA5 total_precipitation yearly files 2016–2025.")
```
