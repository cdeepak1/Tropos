```python
import os
from netCDF4 import Dataset
import pandas as pd
import re

# Directory containing .nc files (current directory)
directory = '.'

# List to store location attributes
locations = []

# Loop through all .nc files in the directory
for filename in os.listdir(directory):
    if filename.endswith('.nc'):
        filepath = os.path.join(directory, filename)
        with Dataset(filepath, 'r') as nc_file:
            # Extract the title attribute
            title = getattr(nc_file, 'title', None)
            if title:
                # Extract the location after "at"
                match = re.search(r'at (.+)', title)
                if match:
                    location_name = match.group(1)
                    locations.append({'File': filename, 'Location': location_name})

# Convert to DataFrame and save to CSV
df_locations = pd.DataFrame(locations)
df_locations.to_csv('locations.csv', index=False)
print("Locations saved to locations.csv")
```

```python
import xarray as xr
import os
import csv

# Configuration
input_dir = 'c:\\Tropos\\Working\\actris\\final_dataset'  # Update this path
output_csv = "output.csv"             # Output CSV filename

# Open CSV file for writing
with open(output_csv, 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    
    # Write CSV header
    writer.writerow(['time', 'location name', 'particle_number_size_distribution'])
    
    # Process each NetCDF file
    for filename in os.listdir(input_dir):
        if filename.endswith('.nc'):
            file_path = os.path.join(input_dir, filename)
            try:
                # Open NetCDF file
                ds = xr.open_dataset(file_path)
                
                # Extract location from title attribute
                title = ds.attrs.get('title', 'Unknown Location')
                location = title.split(' at ')[-1].strip() if ' at ' in title else title
                
                # Extract time and particle data
                times = ds['time'].values
                particle_data = ds['particle_number_size_distribution'].squeeze().values  # Remove singleton dimensions
                
                # Write data to CSV
                for time, value in zip(times, particle_data):
                    writer.writerow([time, location, value])
                
                print(f"Processed: {filename}")
                
            except Exception as e:
                print(f"Error processing {filename}: {str(e)}")
                continue

print(f"Processing complete. Output saved to {output_csv}")
```
